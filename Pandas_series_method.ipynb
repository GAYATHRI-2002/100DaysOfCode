```python
import pandas as pd

# Dictionary containing states/regions and their corresponding values
avg_temp_data = {
    'Kerala': 28,
    'Andhra Pradesh': 28,
    'Tamil Nadu': 31,
    'Karnataka': 28,
    'Delhi': 38,
    'Maharashtra': 33,
    'Rajasthan': 42,
    'Gujarat': 36,
    'Punjab': 35,
    'West Bengal': 32
}



# Creating the Pandas Series from the dictionary
# series_data = pd.Series(data)

# print(series_data)_read = [2,7,4,9]
# s1 = pd.Series(book_read)
```

```python
temp_series = pd.Series(avg_temp_data)
temp_series
```

```python
# Creating a Series with a custom order and an extra state not in the original dictionary
states_list = ['Andhra Pradesh', 'Delhi', 'Tamil Nadu', 'Madhya Pradesh', 'Kerala', 'Karnataka','Rajasthan','Goa']
custom_temp_series = pd.Series(avg_temp_data, index=states_list)
```

```python
custom_temp_series
```

```python
print(custom_temp_series.isna())
```

```python
print(custom_temp_series[custom_temp_series.isna()])
```

```python
print(custom_temp_series.notna())
```
