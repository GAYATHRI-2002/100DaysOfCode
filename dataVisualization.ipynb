```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
```

```python
tips_df = sns.load_dataset('tips')
```

```python
tips_df.head()
```

```python
tips_df.describe()
```

```python
tips_df['day'].describe()
```

```python
# Plot the histogram

plt.hist(tips_df['total_bill'], bins=10, alpha=0.6, edgecolor='black')
plt.title('Distribution of Total Bill')
plt.xlabel('Total Bill Amount ($)')
plt.ylabel('Frequency')
plt.show()
```

```python
# Plot the histogram

plt.hist(tips_df['tip'], bins=10, alpha=0.6, edgecolor='black')
plt.title('Distribution of Tip')
plt.xlabel('Total tip Amount ($)')
plt.ylabel('Frequency')
plt.show()
```

```python
normal_data = np.random.normal(loc=15, scale=3, size=1000)
plt.hist(normal_data, bins=20, alpha=0.6, edgecolor='black')
plt.title('Example of a Normal Distribution')
plt.xlabel('Values')
plt.ylabel('Density')
plt.show()
```
