```python
df = sns.load_dataset('titanic')
```

```python
df.head()
```

```python
df.info()
```

```python
df.isna().sum()
```

```python
df.isnull().sum()
```

```python
missingcounts = df.isnull().sum()
```

```python
plt.bar(missingcounts.index,missingcounts, color = "red")
plt.xlabel('Columns')
plt.ylabel('No of Missing Values')
plt.title('Missing Values in Titanic Dataset per column')
plt.xticks(rotation=60)
plt.show()
```

#handling missing values

```python
df_rows_dropped = df.dropna()
```

```python
print('shape after dropping all rows with missing values ',df_rows_dropped.shape)
```

```python
df_cols_drop = df.dropna(axis=1)
print('remaining columns after dropping missing ones ',df_cols_drop.columns)
```

```python
df_deck_drop = df.drop(columns=['deck'])
print('remaining columns after dropping deck column ',df_deck_drop.columns)
```
